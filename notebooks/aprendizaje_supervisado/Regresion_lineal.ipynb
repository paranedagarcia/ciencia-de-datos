# Regresión Lineal

La **regresión lineal** es una técnica estadística y de machine learning utilizada para modelar la relación entre una variable dependiente (o respuesta) y una o más variables independientes (o predictoras) asumiendo que esta relación es lineal. El objetivo principal es encontrar la línea recta (en el caso de una sola variable independiente) o el hiperplano (en el caso de múltiples variables) que mejor se ajusta a los datos observados, minimizando la suma de los errores cuadrados entre los valores predichos y los valores reales.

En la regresión lineal simple, la relación se expresa mediante la ecuación:

$$
y = \beta_0 + \beta_1 x + \epsilon
$$

donde:
- $y$ es la variable dependiente,
- $x$ es la variable independiente,
- $\beta_0$ es el intercepto,
- $\beta_1$ es el coeficiente de regresión (pendiente),
- $\epsilon$ es el término de error.

En la regresión lineal múltiple, se extiende a varias variables independientes:

$$
y = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots + \beta_n x_n + \epsilon
$$

La regresión lineal es ampliamente utilizada por su simplicidad, interpretabilidad y eficiencia, aunque asume que la relación entre variables es lineal, que los errores son independientes y distribuidos normalmente, y que no existe multicolinealidad entre las variables independientes.

````python
# Ejemplo 1: Regresión lineal simple con scikit-learn
import numpy as np

from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Datos de ejemplo
X = np.array([1, 6, 3, 5, 5,4,1,3,6,7]).reshape(-1, 1)
y = np.array([2, 1, 5, 2, 5,5,6,7,5,8])

# Crear y entrenar el modelo
modelo = LinearRegression()
modelo.fit(X, y)

# Predicciones
y_pred = modelo.predict(X)

# Visualización
plt.scatter(X, y, color='blue', label='Datos reales')
plt.plot(X, y_pred, color='red', label='Recta de regresión')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.title('Regresión lineal simple')
plt.show()
````

**Ejemplo 2: Regresión lineal múltiple**

````python
# Datos de ejemplo para regresión múltiple
X_multi = np.array([[1, 2], [2, 1], [3, 4], [4, 3], [5, 5]])
y_multi = np.array([[2, 4], [1, 6], [7, 3], [1, 6], [1, 7]])

modelo_multi = LinearRegression()
modelo_multi.fit(X_multi, y_multi)

print("Coeficientes:", modelo_multi.coef_)
print("Intercepto:", modelo_multi.intercept_)

# grafica la predicion y los datos reales
y_pred_multi = modelo_multi.predict(X_multi)
plt.scatter(X_multi[:, 0], y_multi[:, 0], color='blue', label='Datos reales y1')
plt.scatter(X_multi[:, 0], y_pred_multi[:, 0], color='red', label='Predicciones y1')
plt.xlabel('X1')
plt.ylabel('y1')
plt.legend()
plt.title('Regresión lineal múltiple y1')
plt.show()

# grafica la predicion y los datos reales para y2
plt.scatter(X_multi[:, 0], y_multi[:, 1], color='blue', label='Datos reales y2')
plt.scatter(X_multi[:, 0], y_pred_multi[:, 1], color='red', label='Predicciones y2')
plt.xlabel('X1')
plt.ylabel('y2')
plt.legend()
plt.title('Regresión lineal múltiple y2')
plt.show()
````

# Regresión lineal OLS (Ordinary Least Squares)

La **regresión lineal OLS (Ordinary Least Squares)** es el método estadístico más utilizado para estimar modelos de regresión lineal. Su objetivo es **encontrar la recta (o hiperplano) que mejor se ajusta a los datos** minimizando la suma de los errores.

---

# 📌 **Definición Analítica**

La regresión lineal busca modelar la relación entre:

* una variable dependiente ( y )
* una o más variables independientes ( x_1, x_2, \ldots, x_p )

El modelo se expresa como:

$
y = \beta_0 + \beta_1 x_1 + \cdots + \beta_p x_p + \varepsilon
$

donde:

* ( $\beta_0$, $\beta_1$, $\ldots$, $\beta_p$ ) son parámetros a estimar,
* ( $varepsilon$ ) es el error aleatorio.

OLS estima los parámetros buscando minimizar:

$
\text{SSE} = \sum_{i=1}^n (y_i - \hat{y}_i)^2
$

Esta es la **suma de los errores al cuadrado** entre los valores observados y predichos.

---

# 📌 **Interpretación Intuitiva**

OLS elige la recta que deja **las distancias verticales más pequeñas posibles** entre los puntos y la recta estimada.

Es el equivalente matemático a decir:

> “Elige el modelo que comete el menor error posible en promedio”.

---

# 📌 **Derivación en forma matricial (visión analítica)**

En forma matricial:

$
\mathbf{y} = \mathbf{X}\beta + \varepsilon
$

La solución OLS es:

$
\hat{\beta} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T \mathbf{y}
$

Este estimador es:

* **insesgado** (en promedio acierta),
* **óptimo** bajo supuestos clásicos (Gauss–Markov),
* tiene varianza mínima entre todos los estimadores lineales insesgados (BLUE).

---

# 📌 **Supuestos fundamentales del OLS**

1. **Linealidad**
   La relación entre predictores y respuesta es lineal.

2. **Independencia de los errores**

3. **Homoscedasticidad**
   Varianza constante de los errores.

4. **Normalidad de los errores**
   (importante para inferencia, no para estimar coeficientes).

5. **No multicolinealidad**
   Los predictores no deben estar fuertemente correlacionados entre sí.

---

# 📊 **Validación (Data Science Framework)**

Después de ajustar un modelo OLS, se evalúan:

### Métricas

* **RMSE**: error promedio.
* **MAE**: error absoluto medio.
* **R²**: proporción de variabilidad explicada.
* **Adjusted R²**: corrige por número de variables.

### Análisis de residuales

* Distribución normal de residuales.
* Independencia.
* Homoscedasticidad (gráfico residuales vs predicción).

````python
"""
Evaluación de Autocorrelación Residual con Durbin-Watson.
Librería: statsmodels.
"""

import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.stattools import durbin_watson

np.random.seed(42)
T = 150  # Número de periodos temporales
t = np.arange(T)

# Variable explicativa: tendencia lineal
X = sm.add_constant(t)

# ==============================================================================
# Escenario A: Residuos Independientes (Ruido Blanco Gaussiano)
# ==============================================================================
epsilon_independiente = np.random.normal(loc=0, scale=5.0, size=T)
y_sin_autocorr = 20.0 + 0.8 * t + epsilon_independiente

modelo_a = sm.OLS(y_sin_autocorr, X).fit()
residuos_a = modelo_a.resid
dw_a = durbin_watson(residuos_a)

# ==============================================================================
# Escenario B: Residuos con Fuerte Autocorrelación Positiva AR(1) inducida
# ==============================================================================
epsilon_ar1 = np.zeros(T)
rho = 0.85  # Coeficiente autorregresivo severo
v = np.random.normal(loc=0, scale=3.0, size=T)

for i in range(1, T):
    epsilon_ar1[i] = rho * epsilon_ar1[i - 1] + v[i]

y_con_autocorr = 20.0 + 0.8 * t + epsilon_ar1

modelo_b = sm.OLS(y_con_autocorr, X).fit()
residuos_b = modelo_b.resid
dw_b = durbin_watson(residuos_b)

# ==============================================================================
# Resumen Diagnóstico
# ==============================================================================
print("=== DIAGNÓSTICO DE DURBIN-WATSON EN MODELOS OLS ===")
print(f"Escenario A (Residuos Independientes): d = {dw_a:.3f}  -> Próximo a 2.0 (Sin autocorrelación)")
print(f"Escenario B (Residuos con AR(1) ρ=0.85): d = {dw_b:.3f}  -> Próximo a 0.0 (Autocorrelación positiva severa)")

# Inspección directa en la tabla de resumen de statsmodels
print("\nExtracto de diagnóstico de Statsmodels (Escenario B):")
print(f"Durbin-Watson reportado en model.summary2(): {modelo_b.summary2().tables[2].iloc[0, 3]}")
````

````python
"""
Evaluación de Normalidad con la Prueba de Jarque-Bera (JB).
Librerías: scipy.stats y statsmodels.
"""

import numpy as np
from scipy import stats
from statsmodels.stats.stattools import jarque_bera

np.random.seed(42)
n_muestras = 1500

# ==============================================================================
# 1. Caso A: Distribución Normal (H0 verdadera)
# ==============================================================================
datos_normales = np.random.normal(loc=0.0, scale=1.0, size=n_muestras)

# Cálculo mediante scipy.stats
res_scipy = stats.jarque_bera(datos_normales)
print("=== CASO A: DATOS CON DISTRIBUCIÓN NORMAL ===")
print(f"Estadístico JB: {res_scipy.statistic:.4f}")
print(f"p-valor:        {res_scipy.pvalue:.4f}  (p >= 0.05 -> No se rechaza normalidad)\n")

# ==============================================================================
# 2. Caso B: Distribución No Normal con Colas Pesadas (t-Student df=3)
# ==============================================================================
datos_t = np.random.standard_t(df=3, size=n_muestras)

# Cálculo mediante statsmodels (desglosa estadístico, p-valor, sesgo y curtosis)
jb_val, p_val, skew, kurtosis = jarque_bera(datos_t)
print("=== CASO B: DATOS CON COLAS PESADAS (t-Student df=3) ===")
print(f"Estadístico JB:     {jb_val:.2f}")
print(f"p-valor:            {p_val:.4e} (p < 0.05 -> Se rechaza normalidad)")
print(f"Asimetría (Skew):   {skew:.3f}   (Normal esperada = 0.0)")
print(f"Curtosis (Kurt):    {kurtosis:.3f}  (Normal esperada = 3.0)")
````

### Comparacion OLS Ridge y Lasso
El siguiente código contrasta el ajuste de **OLS**, **Ridge** y **Lasso** sobre un conjunto de datos multivariable, ilustrando cómo el escalado de características y la sintonización de $\alpha$ modifican los pesos resultantes:

## **Regresión Lasso y Regresión Ridge**
A continuación se presenta una explicación **analítica, comparativa y orientada a ciencia de datos** de **Regresión Lasso** y **Regresión Ridge**, dos extensiones fundamentales de la regresión lineal OLS cuando existe **colinealidad**, **sobreajuste**, o muchas variables.


### 🧠 **Contexto: ¿Por qué surgen Lasso y Ridge?**

En OLS, el objetivo es minimizar:

$
\sum (y_i - \hat{y}_i)^2
$

Pero cuando:

* hay **muchas variables**,
* existe **multicolinealidad**,
* o el modelo presenta **overfitting**,

OLS pierde estabilidad porque la matriz (X^TX) se vuelve mal condicionada.

Para controlar esto, se agrega una **penalización** al tamaño de los coeficientes, lo que estabiliza el modelo.

---

## 🟦 **Regresión Ridge (L2 Regularization)**

**Definición**

Ridge agrega una penalización L2 a la función de pérdida:

$
\underset{\beta}{\min} \left[ \sum_{i=1}^n (y_i - \hat{y}*i)^2 + \lambda \sum*{j=1}^p \beta_j^2 \right]
$

donde:

* ( \lambda \ge 0 ) controla la fuerza de regularización,
* cuando ( \lambda = 0 ) → es OLS,
* cuando ( \lambda → \infty ) → coeficientes tienden a 0 pero **no llegan a ser exactamente 0**.

**Qué logra Ridge**

* Reduce varianza del modelo.
* Estabiliza los coeficientes cuando hay alta colinealidad.
* Hace que pesos grandes se “encojan” (shrinkage).
* No realiza selección de variables.

**Interpretación geométrica**

Penaliza grandes coeficientes usando una esfera L2, lo que limita la magnitud total.

---

## 🟪 **Regresión Lasso (L1 Regularization)**

**Definición**

Lasso agrega una penalización L1:

$
\underset{\beta}{\min} \left[ \sum_{i=1}^n (y_i - \hat{y}*i)^2 + \lambda \sum*{j=1}^p |\beta_j| \right]
$

**Diferencias clave con Ridge**

* Penaliza con valores absolutos.
* Puede **llevar coeficientes exactamente a cero**.
* Actúa como un método de **selección automática de variables**.

**Cuándo usarlo**

* Cuando existen muchas variables irrelevantes.
* Para reducir dimensionalidad.
* Cuando se busca interpretabilidad.

**Interpretación geométrica**

Lasso utiliza un polígono (diamante) L1; sus esquinas favorecen soluciones en 0, por eso elimina variables.

---

# 📊 **Comparación técnica**

| Aspecto                     | Ridge (L2)       | Lasso (L1)                             |       |   |
| --------------------------- | ---------------- | -------------------------------------- | ----- | - |
| Penalización                | $ \sum \beta^2 $ | ( \sum                                 | \beta | ) |
| Efecto                      | Reduce magnitud  | Puede anular coeficientes              |       |   |
| Selección de variables      | ❌ No             | ✔ Sí                                   |       |   |
| Manejo de multicolinealidad | ✔ Excelente      | ✔ Pero selecciona 1 variable del grupo |       |   |
| Estabilidad                 | Muy estable      | Menos estable cuando p >> n            |       |   |
| Interpretabilidad           | Media            | Alta                                   |       |   |
| Uso típico                  | Predicción       | Feature selection                      |       |   |

---

# 📐 **Hipótesis Estadísticas y Data Science Framework**

Ambos son métodos de **regularización**, que:

* reducen varianza del modelo,
* mitigan el sobreajuste,
* mejoran la generalización,
* sesgan los coeficientes hacia cero, añadiendo *bias* controlado.

Este es el clásico **bias–variance tradeoff**.

El parámetro clave es:

$
\lambda
$

que se ajusta normalmente por **validación cruzada (CV)**.

---


---

## 📌 **¿Cómo elegir entre Lasso y Ridge?**

### ✔ Usa **Ridge** cuando:

* Todas las variables aportan algo.
* Hay multicolinealidad fuerte.
* El objetivo es maximizar precisión predictiva.

### ✔ Usa **Lasso** cuando:

* Hay muchas variables irrelevantes.
* Necesitas seleccionar features.
* Buscas interpretabilidad o simplificación del modelo.

### ✔ Híbrido: **Elastic Net**

Cuando quieres lo mejor de ambos:

$
\alpha L1 + (1-\alpha) L2
$

---

**Ejemplo visual**

Ridge → “encoge” coeficientes
Lasso → “encoge y elimina”

```
coeficiente_OLS = 2.5
coeficiente_Ridge ≈ 1.4
coeficiente_Lasso = 0.0
```

---

## ✔ Conclusión

**Ridge y Lasso son extensiones de la regresión lineal OLS que aplican regularización para mejorar la generalización del modelo.**

* **Ridge estabiliza y reduce varianza.**
* **Lasso selecciona variables automáticamente.**

Ambos son pilares en ML aplicado a datos tabulares, especialmente en problemas con muchas variables o colinealidad.

---



````python
"""
Comparativa OLS vs. Ridge vs. Lasso.
Demostración de contracción de pesos y selección de características.
"""

import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score

# 1. Carga y partición de datos
housing = fetch_california_housing(as_frame=True)
X, y = housing.data, housing.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# 2. Preprocesamiento: Estandarización obligatoria para modelos regularizados
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 3. Ajuste de Mínimos Cuadrados Ordinarios (OLS)
ols_model = LinearRegression()
ols_model.fit(X_train_scaled, y_train)
y_pred_ols = ols_model.predict(X_test_scaled)

# 4. Ajuste de Regresión Ridge (Penalización L2)
ridge_model = Ridge(alpha=100.0, random_state=42)
ridge_model.fit(X_train_scaled, y_train)
y_pred_ridge = ridge_model.predict(X_test_scaled)

# 5. Ajuste de Regresión Lasso (Penalización L1)
lasso_model = Lasso(alpha=0.05, random_state=42)
lasso_model.fit(X_train_scaled, y_train)
y_pred_lasso = lasso_model.predict(X_test_scaled)

# 6. Inspección de Coeficientes y Dispersión
df_coeficientes = pd.DataFrame({
    'Característica': X.columns,
    'OLS (Sin penalizar)': ols_model.coef_,
    'Ridge (L2)': ridge_model.coef_,
    'Lasso (L1)': lasso_model.coef_
})

print("=== MAGNITUD DE COEFICIENTES ESTIMADOS ===")
print(df_coeficientes.round(4).to_string(index=False))

print("\n=== EVALUACIÓN EN CONJUNTO DE PRUEBA (TEST SET) ===")
print(f"OLS   -> RMSE: {np.sqrt(mean_squared_error(y_test, y_pred_ols)):.4f} | R²: {r2_score(y_test, y_pred_ols):.4f}")
print(f"Ridge -> RMSE: {np.sqrt(mean_squared_error(y_test, y_pred_ridge)):.4f} | R²: {r2_score(y_test, y_pred_ridge):.4f}")
print(f"Lasso -> RMSE: {np.sqrt(mean_squared_error(y_test, y_pred_lasso)):.4f} | R²: {r2_score(y_test, y_pred_lasso):.4f}")

# Contabilizar características anuladas por Lasso
n_anuladas = np.sum(lasso_model.coef_ == 0.0)
print(f"\nVariables anuladas exactamente a cero por Lasso: {n_anuladas} de {X.shape}")
````

````python
"""
Regresión Lineal y Regularización (OLS vs. Ridge vs. Lasso).
Demostración del encogimiento y dispersión de coeficientes.
"""

import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score

# 1. Ingesta y partición de datos
california = fetch_california_housing(as_frame=True)
X = california.data
y = california.target

# Dividimos en 75% entrenamiento y 25% prueba para evaluar generalización
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# 2. Preprocesamiento: Estandarización de variables
# REGLA RIGUROSA: En modelos regularizados (L1/L2), las variables deben estar
# en la misma escala (media 0, varianza 1); de lo contrario, las características
# con valores absolutos mayores dominarían injustamente la penalización.
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ==============================================================================
# 3. Ajuste de Modelos
# ==============================================================================

# A. Mínimos Cuadrados Ordinarios (OLS)
ols_model = LinearRegression()
ols_model.fit(X_train_scaled, y_train)
y_pred_ols = ols_model.predict(X_test_scaled)

# B. Regresión Ridge (Penalización L2)
# alpha regula la fuerza de contracción (lambda en la formulación teórica)
ridge_model = Ridge(alpha=100.0, random_state=42)
ridge_model.fit(X_train_scaled, y_train)
y_pred_ridge = ridge_model.predict(X_test_scaled)

# C. Regresión Lasso (Penalización L1)
# alpha regula el umbral de dispersión y selección de variables
lasso_model = Lasso(alpha=0.08, max_iter=10000, random_state=42)
lasso_model.fit(X_train_scaled, y_train)
y_pred_lasso = lasso_model.predict(X_test_scaled)

# ==============================================================================
# 4. Inspección de Coeficientes y Dispersión
# ==============================================================================
df_coeficientes = pd.DataFrame({
    'Variable': X.columns,
    'OLS (Sin regularizar)': ols_model.coef_,
    'Ridge (L2)': ridge_model.coef_,
    'Lasso (L1)': lasso_model.coef_
})

print("=== MAGNITUD DE COEFICIENTES DE REGRESIÓN ===")
print(df_coeficientes.round(4).to_string(index=False))

# Conteo de coeficientes forzados a exactamente cero
zeros_ols = np.sum(ols_model.coef_ == 0.0)
zeros_ridge = np.sum(ridge_model.coef_ == 0.0)
zeros_lasso = np.sum(lasso_model.coef_ == 0.0)

print("\n=== RECUENTO DE VARIABLES ANULADAS (SPARSITY) ===")
print(f"OLS:   {zeros_ols} variables en 0 de {X.shape}")
print(f"Ridge: {zeros_ridge} variables en 0 de {X.shape}")
print(f"Lasso: {zeros_lasso} variables en 0 de {X.shape} (Selección automática activa)")

# ==============================================================================
# 5. Evaluación de Rendimiento en el Conjunto de Prueba (Test Set)
# ==============================================================================
def evaluar_rendimiento(nombre, y_real, y_pred):
    rmse = np.sqrt(mean_squared_error(y_real, y_pred))
    r2 = r2_score(y_real, y_pred)
    print(f"{nombre:<8} | RMSE: {rmse:.4f} | R²: {r2:.4f}")

print("\n=== EVALUACIÓN EN CONJUNTO DE PRUEBA (TEST SET) ===")
evaluar_rendimiento("OLS", y_test, y_pred_ols)
evaluar_rendimiento("Ridge", y_test, y_pred_ridge)
evaluar_rendimiento("Lasso", y_test, y_pred_lasso)
````

### Progresion de diabetes
flujo de trabajo:
1. Carga los datos y genera interacciones polinómicas de segundo grado para inducir multicolinealidad.
2. Separa un conjunto de prueba (*held-out test set*) del 20% que los modelos **nunca verán** durante el entrenamiento.
3. Estandariza las características de forma rigurosa (`StandardScaler`) ajustándolo solo en entrenamiento para prevenir fuga de datos (*data leakage*).
4. Utiliza validación cruzada de 10 pliegues (`RidgeCV`, `LassoCV`, `ElasticNetCV`) para encontrar el valor óptimo del hiperparámetro $\alpha$.
5. Evalúa el desempeño en datos de prueba no vistos mediante **RMSE** (Raíz del Error Cuadrático Medio), **MAE** (Error Absoluto Medio) y el coeficiente de determinación **$R^2$**.

````python
"""
Comparación Rigurosa de Regularización Lineal
Dataset: Diabetes Progression (con expansión polinómica de características)
"""

import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV, ElasticNetCV
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# ==============================================================================
# 1. Carga y Expansión de Características
# ==============================================================================
diabetes = load_diabetes(as_frame=True)
X_raw = diabetes.data
y = diabetes.target.values

# Inducimos multicolinealidad e interacciones no lineales de 2do orden
# (Esto expande las 10 características a 65 columnas: x_i * x_j)
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X_raw)
feature_names = poly.get_feature_names_out(X_raw.columns)

print(f"Dimensiones iniciales: {X_raw.shape}")
print(f"Dimensiones tras expansión polinómica: {X_poly.shape}")

# ==============================================================================
# 2. Partición de Datos (Train / Test Split)
# ==============================================================================
X_train, X_test, y_train, y_test = train_test_split(
    X_poly, y, test_size=0.20, random_state=42
)

# ==============================================================================
# 3. Preprocesamiento: Estandarización Estricta
# ==============================================================================
# REGLA CRÍTICA: Ajustar (fit) el escalador exclusivamente con el conjunto de entrenamiento
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ==============================================================================
# 4. Entrenamiento y Sintonización de Hiperparámetros vía Validación Cruzada
# ==============================================================================
alphas_grid = np.logspace(-3, 3, 200)

# A. Mínimos Cuadrados Ordinarios (Sin Regularización)
ols = LinearRegression()
ols.fit(X_train_scaled, y_train)

# B. Ridge (Regularización L2) con CV
ridge = RidgeCV(alphas=alphas_grid, cv=10, scoring='neg_mean_squared_error')
ridge.fit(X_train_scaled, y_train)

# C. Lasso (Regularización L1) con CV
lasso = LassoCV(alphas=alphas_grid, cv=10, max_iter=20000, random_state=42)
lasso.fit(X_train_scaled, y_train)

# D. ElasticNet (Regularización Híbrida L1 + L2) con CV
l1_ratios = [0.1, 0.3, 0.5, 0.7, 0.9, 0.95, 0.99]
elastic = ElasticNetCV(alphas=alphas_grid, l1_ratio=l1_ratios, cv=10, max_iter=20000, random_state=42)
elastic.fit(X_train_scaled, y_train)

# ==============================================================================
# 5. Evaluación de Desempeño en el Conjunto de Prueba (Test Set)
# ==============================================================================
modelos = {
    "OLS (Sin regularizar)": ols,
    f"Ridge (L2, α={ridge.alpha_:.4f})": ridge,
    f"Lasso (L1, α={lasso.alpha_:.4f})": lasso,
    f"ElasticNet (α={elastic.alpha_:.4f}, l1_ratio={elastic.l1_ratio_})": elastic
}

resultados = []

for nombre, modelo in modelos.items():
    y_pred = modelo.predict(X_test_scaled)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    
    # Conteo de coeficientes descartados (exactamente iguales a 0)
    coefs = modelo.coef_
    n_ceros = np.sum(np.isclose(coefs, 0.0, atol=1e-5))
    n_activos = len(coefs) - n_ceros
    
    resultados.append({
        "Modelo": nombre,
        "RMSE (Menor es mejor)": round(rmse, 3),
        "MAE (Menor es mejor)": round(mae, 3),
        "R² (Mayor es mejor)": round(r2, 4),
        "Variables Activas": f"{n_activos} / {len(coefs)}",
        "Variables a Cero": n_ceros
    })

df_resultados = pd.DataFrame(resultados)
print("\n=== TABLA COMPARATIVA DE DESEMPEÑO (TEST SET) ===")
print(df_resultados.to_string(index=False))
````

````python
# Implementación en Python
"""
Predicción de Bienes Raíces y Regularización en Python.
Evalúa OLS, Ridge, ElasticNet y Lasso con gráficos de diagnóstico en cuadrícula 2x2.
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, RidgeCV, LassoCV, ElasticNetCV
from sklearn.metrics import mean_squared_error, r2_score

# ==============================================================================
# 1. Simulación del Conjunto de Datos de Bienes Raíces (N = 500 propiedades)
# ==============================================================================
np.random.seed(42)
N = 500

# Variables estructurales reales
superficie_m2 = np.random.uniform(50, 320, size=N)        # Superficie construida
habitaciones = np.random.choice([1, 2, 3, 4, 5], size=N, p=[0.1, 0.25, 0.35, 0.2, 0.1])
distancia_centro_km = np.random.uniform(1.0, 25.0, size=N) # Distancia al centro financiero
antiguedad_anios = np.random.uniform(0, 45, size=N)       # Antigüedad de la edificación
calidad_acabados = np.random.uniform(1, 10, size=N)       # Índice de calidad 1-10

# Precio real de mercado (en miles de USD - $K) con una ligera convexidad de lujo
precio_base = (
    85.0 
    + 1.85 * superficie_m2 
    + 14.0 * habitaciones 
    - 4.2 * distancia_centro_km 
    - 1.1 * antiguedad_anios 
    + 9.5 * calidad_acabados
    + 0.003 * (superficie_m2 ** 2)
)
ruido_mercado = np.random.normal(0, 22.0, size=N)
y = precio_base + ruido_mercado

# Atributos redundantes / colineales (ingesta de portales duplicados)
superficie_sqft = superficie_m2 * 10.7639 + np.random.normal(0, 0.5, size=N) # Colineal a m²
banios_aprox = 0.6 * habitaciones + np.random.normal(0, 0.3, size=N)
tiempo_traslado_min = distancia_centro_km * 2.8 + np.random.normal(0, 1.5, size=N) # Colineal a distancia
anio_construccion = 2026 - antiguedad_anios # Colineal inversa perfecta

# Atributos espurios (ruido ambiental sin capacidad predictiva)
ruido_1 = np.random.normal(50, 15, size=N)
ruido_2 = np.random.uniform(100, 200, size=N)
ruido_3 = np.random.choice([0, 1], size=N)

# Matriz de diseño con 12 características
df_inmuebles = pd.DataFrame({
    'Superficie_m2': superficie_m2,
    'Habitaciones': habitaciones,
    'Distancia_Centro_km': distancia_centro_km,
    'Antiguedad_anios': antiguedad_anios,
    'Calidad_Acabados': calidad_acabados,
    'Superficie_sqft': superficie_sqft,
    'Banios_estimados': banios_aprox,
    'Tiempo_traslado_min': tiempo_traslado_min,
    'Anio_construccion': anio_construccion,
    'Ruido_sensor_1': ruido_1,
    'Ruido_sensor_2': ruido_2,
    'Ruido_indicador_lote': ruido_3
})

X = df_inmuebles.values

# ==============================================================================
# 2. Partición Rigurosa y Preprocesamiento
# ==============================================================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Estandarización Z-score ajustada exclusivamente sobre entrenamiento
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ==============================================================================
# 3. Ajuste y Optimización de Hiperparámetros vía Validación Cruzada (5-Folds)
# ==============================================================================
ols = LinearRegression().fit(X_train_scaled, y_train)
ridge = RidgeCV(cv=5).fit(X_train_scaled, y_train)
lasso = LassoCV(cv=5, random_state=42, max_iter=10000).fit(X_train_scaled, y_train)
elastic = ElasticNetCV(
    cv=5, l1_ratio=[0.1, 0.5, 0.7, 0.9], random_state=42, max_iter=10000
).fit(X_train_scaled, y_train)

modelos_dict = {
    'OLS (Sin regularizar)': ols,
    f'Ridge (L2, α={ridge.alpha_:.2f})': ridge,
    f'ElasticNet (α={elastic.alpha_:.2f})': elastic,
    f'Lasso (L1, α={lasso.alpha_:.2f})': lasso
}

# ==============================================================================
# 4. Evaluación de Métricas de Generalización
# ==============================================================================
nombres, rmses, r2s, activos, anulados = [], [], [], [], []
predicciones = {}

for nombre, m in modelos_dict.items():
    pred = m.predict(X_test_scaled)
    predicciones[nombre] = pred
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    r2 = r2_score(y_test, pred)
    
    coefs = m.coef_
    n_ceros = np.sum(np.isclose(coefs, 0.0, atol=1e-3))
    n_activos = X.shape[1] - n_ceros
    
    nombres.append(nombre)
    rmses.append(rmse)
    r2s.append(r2 * 100)
    activos.append(n_activos)
    anulados.append(n_ceros)
    
    print(f"{nombre:<32} | RMSE: ${rmse:.2f}K | R²: {r2*100:.2f}% | Variables activas: {n_activos}/12")

# ==============================================================================
# 5. Generación de Gráficos Diagnósticos (2x2 con Fondo Blanco Plano)
# ==============================================================================
fig, axes = plt.subplots(2, 2, figsize=(15, 11), facecolor='white')
fig.suptitle(
    'Evaluación de Modelos Lineales para Tasación Inmobiliaria (N=500)',
    fontsize=16, fontweight='bold', y=0.98
)

paleta = ['#d55e00', '#0072b2', '#e69f00', '#009e73']

# Panel 1: Error Cuadrático Medio (RMSE)
ax1 = axes[0, 0]
ax1.set_facecolor('white')
barras_rmse = ax1.bar(nombres, rmses, color=paleta, width=0.52)
ax1.set_title('1. Error de Predicción (RMSE en Test Set)', fontsize=12, fontweight='bold', pad=10)
ax1.set_ylabel('RMSE ($K USD)', fontsize=11)
ax1.set_ylim(0, max(rmses) * 1.25)
ax1.grid(axis='y', linestyle=':', alpha=0.6)
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)
for b, val in zip(barras_rmse, rmses):
    ax1.text(b.get_x() + b.get_width()/2, val + 0.6, f'${val:.2f}K', ha='center', va='bottom', fontweight='bold', fontsize=10)

# Panel 2: Coeficiente de Determinación (R²)
ax2 = axes[0, 1]
ax2.set_facecolor('white')
barras_r2 = ax2.bar(nombres, r2s, color=paleta, width=0.52)
ax2.set_title('2. Varianza Explicada (R² en Test Set)', fontsize=12, fontweight='bold', pad=10)
ax2.set_ylabel('R² (%)', fontsize=11)
ax2.set_ylim(85, 100)
ax2.grid(axis='y', linestyle=':', alpha=0.6)
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
for b, val in zip(barras_r2, r2s):
    ax2.text(b.get_x() + b.get_width()/2, val + 0.3, f'{val:.1f}%', ha='center', va='bottom', fontweight='bold', fontsize=10)

# Panel 3: Principio de Parsimonia
ax3 = axes[1, 0]
ax3.set_facecolor('white')
idx = np.arange(len(nombres))
b_act = ax3.bar(idx, activos, width=0.52, label='Variables Activas (w ≠ 0)', color='#0072b2')
b_anu = ax3.bar(idx, anulados, width=0.52, bottom=activos, label='Variables Descartadas a Cero (w = 0)', color='#cccccc')
ax3.set_title('3. Parsimonia: Variables Activas vs. Anuladas', fontsize=12, fontweight='bold', pad=10)
ax3.set_ylabel('Número de Atributos (Total: 12)', fontsize=11)
ax3.set_xticks(idx)
ax3.set_xticklabels(nombres, rotation=10, ha='right', fontsize=9.5)
ax3.set_ylim(0, 15)
ax3.grid(axis='y', linestyle=':', alpha=0.6)
ax3.spines['top'].set_visible(False)
ax3.spines['right'].set_visible(False)
ax3.legend(frameon=True, facecolor='white', loc='upper right')
for i, (act, anu) in enumerate(zip(activos, anulados)):
    ax3.text(i, act/2, f'{act}', ha='center', va='center', color='white', fontweight='bold')
    if anu > 0:
        ax3.text(i, act + anu/2, f'{anu}', ha='center', va='center', color='#333333', fontweight='bold')

# Panel 4: Alineación Real vs. Predicho (OLS vs. Lasso)
ax4 = axes[1, 1]
ax4.set_facecolor('white')
p_ols = predicciones['OLS (Sin regularizar)']
nombre_lasso = f'Lasso (L1, α={lasso.alpha_:.2f})'
p_lasso = predicciones[nombre_lasso]
rmse_ols = rmses[nombres.index('OLS (Sin regularizar)')]
rmse_lasso = rmses[nombres.index(nombre_lasso)]

ax4.scatter(y_test, p_ols, color='#d55e00', alpha=0.5, s=35, label=f'OLS (RMSE: ${rmse_ols:.2f}K)')
ax4.scatter(y_test, p_lasso, color='#009e73', alpha=0.75, s=35, label=f'Lasso (RMSE: ${rmse_lasso:.2f}K)')

lim_inf = min(y_test.min(), p_ols.min(), p_lasso.min()) - 20
lim_sup = max(y_test.max(), p_ols.max(), p_lasso.max()) + 20
ax4.plot([lim_inf, lim_sup], [lim_inf, lim_sup], color='#333333', linestyle='--', linewidth=1.5, label='Alineación Ideal (1:1)')
ax4.set_title('4. Alineación: Valor Real vs. Predicción', fontsize=12, fontweight='bold', pad=10)
ax4.set_xlabel('Precio Real de Venta ($K USD)', fontsize=11)
ax4.set_ylabel('Precio Predicho por el Modelo ($K USD)', fontsize=11)
ax4.set_xlim(lim_inf, lim_sup)
ax4.set_ylim(lim_inf, lim_sup)
ax4.grid(True, linestyle=':', alpha=0.6)
ax4.spines['top'].set_visible(False)
ax4.spines['right'].set_visible(False)
ax4.legend(frameon=True, facecolor='white', loc='upper left')

plt.tight_layout()
plt.show()
````
