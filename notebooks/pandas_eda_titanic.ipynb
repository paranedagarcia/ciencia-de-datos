## 3. Detección y Análisis de Valores Nulos (Missing Values)

En datasets del mundo real, es estándar encontrar datos faltantes. Calculamos cuántos valores nulos existen por columna y qué porcentaje del total representan.


```python
# Conteo y porcentaje de valores nulos por columna
nulos_conteo = df.isna().sum()
nulos_porcentaje = (df.isna().sum() / len(df)) * 100

resumen_nulos = pd.DataFrame({
    'Total Nulos': nulos_conteo,
    'Porcentaje (%)': nulos_porcentaje.round(2)
})

# Filtrar únicamente las columnas que contienen datos faltantes
resumen_nulos[resumen_nulos['Total Nulos'] > 0]
```

**Interpretación:**
* La columna `Cabin` tiene más del 77% de datos faltantes (difícil de imputar).
* `Age` tiene cerca del 20% de valores faltantes (177 pasajeros).
* `Embarked` tiene solo 2 valores nulos (fácilmente imputables con la moda).

## 6. Agrupaciones con `groupby` y Agregaciones (`agg`)

El método `groupby` implementa la estrategia fundamental de análisis de datos:
1. **Split (Dividir):** Separar los datos en grupos basados en criterios específicos.
2. **Apply (Aplicar):** Ejecutar una función de agregación (media, suma, conteo, mediana) en cada grupo.
3. **Combine (Combinar):** Unir los resultados en una nueva estructura de datos.

---

### 📝 Ejercicio 3: Tasa de Supervivencia por Género
Calcula la tasa global de supervivencia desglosada por sexo del pasajero.


```python
# SOLUCIÓN EJERCICIO 3:
supervivencia_sexo = df.groupby('Sex')['Survived'].agg(
    Total_Pasajeros='count',
    Sobrevivientes='sum',
    Tasa_Supervivencia='mean'
)

# Convertir la tasa a formato porcentual
supervivencia_sexo['Tasa_Supervivencia (%)'] = (supervivencia_sexo['Tasa_Supervivencia'] * 100).round(2)
supervivencia_sexo
```

### 📝 Ejercicio 4: Análisis Multivariable por Clase y Género
Calcula cuántos pasajeros viajaban y cuál fue la tasa de supervivencia agrupando simultáneamente por `Pclass` y `Sex`.


```python
# SOLUCIÓN EJERCICIO 4:
supervivencia_clase_sexo = df.groupby(['Pclass', 'Sex'])['Survived'].agg(
    Pasajeros='count',
    Tasa_Supervivencia='mean'
).round(4)

supervivencia_clase_sexo['Tasa (%)'] = (supervivencia_clase_sexo['Tasa_Supervivencia'] * 100).round(2)
supervivencia_clase_sexo
```

### 📝 Ejercicio 5: Tarifas y Edades según la Clase de Boleto
Calcula para cada clase (`Pclass`) el promedio, mediana, mínimo y máximo de la tarifa (`Fare`), y la edad promedio (`Age`).


```python
# SOLUCIÓN EJERCICIO 5:
metricas_clase = df.groupby('Pclass').agg({
    'Fare': ['mean', 'median', 'min', 'max'],
    'Age': ['mean', 'median']
}).round(2)

metricas_clase
```

## 7. Tablas de Contingencia (`pd.crosstab`) y Tablas Dinámicas (`pivot_table`)

Las tablas cruzadas permiten comparar la distribución de frecuencias entre variables categóricas, mientras que las tablas dinámicas permiten agregar métricas con márgenes de totales.

---

### 📝 Ejercicio 6: Tabla Cruzada de Clase vs Supervivencia con Porcentajes
Crea una tabla cruzada entre `Pclass` y `Survived` que muestre el porcentaje de supervivencia por fila (normalizado).


```python
# SOLUCIÓN EJERCICIO 6:
# normalize='index' calcula los porcentajes por fila (cada clase suma 100%)
crosstab_clase_superv = pd.crosstab(
    df['Pclass'], 
    df['Survived'], 
    normalize='index', 
    margins=True, 
    margins_name='Total'
) * 100

crosstab_clase_superv.columns = ['Fallecidos (%)', 'Sobrevivientes (%)']
crosstab_clase_superv.round(2)
```

### 📝 Ejercicio 7: Tabla Dinámica Completa (`pivot_table`)
Construye una tabla dinámica que muestre la tarifa promedio pagada (`Fare`) desglosada por `Pclass` (en filas) y `Embarked` (en columnas), incluyendo los subtotales (`margins=True`).


```python
# SOLUCIÓN EJERCICIO 7:
tabla_dinamica_tarifas = df.pivot_table(
    values='Fare',
    index='Pclass',
    columns='Embarked',
    aggfunc='mean',
    margins=True,
    margins_name='Promedio General'
).round(2)

tabla_dinamica_tarifas
```

## 8. Ingeniería de Características Básica (Feature Engineering)

A menudo, las columnas originales pueden combinarse para extraer nueva información analítica valiosa.

---

### 📝 Ejercicio 8: Tamaño de Familia y Tipo de Pasajero
Crea una nueva columna llamada `FamilySize` que sume los hermanos/cónyuges (`SibSp`), padres/hijos (`Parch`) más el propio pasajero (`+ 1`).
Luego, clasifica a los pasajeros en:
* `'Solitario'` (si FamilySize == 1)
* `'Familia Pequeña'` (si 2 <= FamilySize <= 4)
* `'Familia Grande'` (si FamilySize >= 5)


```python
# SOLUCIÓN EJERCICIO 8:
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# Crear categorías de familia
def categorizar_familia(tamano):
    if tamano == 1:
        return 'Solitario'
    elif tamano <= 4:
        return 'Familia Pequeña'
    else:
        return 'Familia Grande'

df['TipoFamilia'] = df['FamilySize'].apply(categorizar_familia)

# Analizar la tasa de supervivencia según el tipo de familia
df.groupby('TipoFamilia')['Survived'].agg(
    Total='count',
    Tasa_Supervivencia='mean'
).round(4)
```

### 📝 Ejercicio 9: Segmentación por Grupos de Edad con `pd.cut`
Utiliza `pd.cut` para dividir la edad en 4 rangos etarios:
* Infante (0 - 12 años)
* Joven (12 - 25 años)
* Adulto (25 - 50 años)
* Mayor (50 - 100 años)


```python
# SOLUCIÓN EJERCICIO 9:
limites_edad = [0, 12, 25, 50, 100]
etiquetas_edad = ['Infante (0-12)', 'Joven (12-25)', 'Adulto (25-50)', 'Mayor (50+)']

df['RangoEtario'] = pd.cut(df['Age'], bins=limites_edad, labels=etiquetas_edad, right=False)

# Supervivencia por rango etario y sexo (especificando observed=False para evitar advertencias en pandas moderno)
supervivencia_edad_sexo = df.pivot_table(
    values='Survived',
    index='RangoEtario',
    columns='Sex',
    aggfunc=['count', 'mean'],
    observed=False
).round(3)

supervivencia_edad_sexo
```
